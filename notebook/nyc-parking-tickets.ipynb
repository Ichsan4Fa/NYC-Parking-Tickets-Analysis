# EDA

## NaN Identification
We will explore this 2017 dataset to calculate NaN values inside each columns.

```python
nan_df = df_2017.isna()
nans_gr = {}
for col in df_2017.columns:
    current_nan = nan_df[col].sum()
    try:
        nans_gr[current_nan].append(col)
    except:
        nans_gr[current_nan]=[col]
del nan_df; x=gc.collect()

for k,v in nans_gr.items():
    print('####### NAN count =',k)
    print(v)
```

```python
nan_df_2017 = df_2017.isna()
top_10_NaN = {}
for col in df_2017.columns:
    current_nan = nan_df_2017[col].sum()
    try:
        top_10_NaN[current_nan].append(col)
    except:
        top_10_NaN[current_nan]=[col]

top_10_NaN = dict(sorted(top_10_NaN.items(), key=lambda item: item[1]))

#Bar Chart 10 most NaN features
sn.barplot(x='feature',y='NaN Counts')

plt.show()
```
